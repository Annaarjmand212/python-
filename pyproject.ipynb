```python
class Car:
    def __init__(self, Brand, model, year, price):
        self.Brand = Brand
        self.model = model
        self.year = year
        self.price = price

    def information(self):
        print(f"This is a {self.model} from Brand {self.Brand} model {self.year} . The price of this car is {self.price}")

class Car2(Car):
    def __init__(self, Brand, model, year, price):
        Car.__init__(self, Brand, model, year, price)


c1 = Car("BMW", "2232", "2026", "250k")
c2 = Car2("sportage", "239", "2026", "300k")
print(c1.Brand)
print(c1.model)
print(c1.year)
print(c1.price)
print(c2.Brand)
print(c2.model)
print(c2.year)
print(c2.price)
c1.information()
c2.information()
```

```python
from abc import ABC, abstractmethod


# Abstraction
class BankAccount(ABC):

    def __init__(self, username, account_number, password, balance=0):
        self.username = username
        self.account_number = account_number

        # Encapsulation
        self.__password = password
        self.__balance = balance


    def check_password(self, password):
        return self.__password == password


    def deposit(self, amount):
        self.__balance += amount
        print(f"{amount} deposited successfully")


    def withdraw(self, amount, password):

        if self.check_password(password):

            if amount <= self.__balance:
                self.__balance -= amount
                print(f"{amount} withdrawn successfully")

            else:
                print("Not enough money")

        else:
            print("Wrong password")


    def show_balance(self):
        print(f"Balance: {self.__balance}")


    def transfer(self, other_account, amount, password):

        if self.check_password(password):

            if amount <= self.__balance:
                self.__balance -= amount
                other_account.deposit(amount)

                print("Transfer successful")

            else:
                print("Not enough money")

        else:
            print("Wrong password")


    # Abstraction
    @abstractmethod
    def show_account_type(self):
        pass




# Inheritance
class SavingAccount(BankAccount):

    def __init__(self, username, account_number, password, balance, interest):
        super().__init__(username, account_number, password, balance)
        self.interest = interest


    # Polymorphism
    def show_account_type(self):
        print("This is a Saving Account")



class NormalAccount(BankAccount):

    # Polymorphism
    def show_account_type(self):
        print("This is a Normal Account")






account1 = NormalAccount(
    "Ali",
    12345,
    "1111",
    5000
)


account2 = SavingAccount(
    "Sara",
    67890,
    "2222",
    10000,
    5
)



account1.show_account_type()
account2.show_account_type()


account1.deposit(2000)

account1.withdraw(1000, "1111")


account1.transfer(account2, 3000, "1111")


account1.show_balance()
account2.show_balance()
```
